```python
import pandas as pd

df = pd.read_csv(
    "mymoviedb.csv",
    engine="python",
    on_bad_lines="skip"   # skips malformed/broken rows
)

df.head()
```

```python
df.shape
```

```python
df_clean = df.dropna()
```

```python
df_clean.shape
```

```python
df_clean.loc[:, "Release_Date"] = pd.to_datetime(df_clean["Release_Date"], errors="coerce")
df_clean.loc[:, "Release_Year"] = df_clean["Release_Date"].dt.year
```

```python
df_clean.head()
```

```python
df_clean = df_clean.drop(columns=["Release_Date"])
```

```python
df_clean.head()
```

```python
df_clean = df_clean.rename(columns={"Release_Year": "Release_Date"})
```

```python
import re

# Create clean Name column
df_clean["Name"] = (
    df_clean["Title"]
    .str.lower()                          # lowercase
    .apply(lambda x: re.sub(r'[^a-z]+', '', x))   # keep only a–z
)
```

```python
df_clean["Original_Language"].unique()
```

```python
df_clean.to_csv("cleaned_movies.csv", index=False)
```
